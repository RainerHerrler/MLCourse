# Logistic Regression on Cancer Data

**Data creation**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

cancer = load_breast_cancer()

X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, stratify=cancer.target, random_state=42)
print(X_train.shape)
print(X_test.shape)
```

```python
cancer.feature_names
```

**Train the model**

```python
from sklearn.linear_model import LogisticRegression

logreg = LogisticRegression(C=1, max_iter=10000)
logreg.fit(X_train,y_train)
print("accuracy on train-set: {:.4f}".format(logreg.score(X_train, y_train)))
print("accuracy on test-set : {:.4f}".format(logreg.score(X_test, y_test)))

#print internal data
print("Coeffizienten for every feature: ", logreg.coef_.shape)
print("Intrecept value: ", logreg.intercept_.shape)
```

<B>Trying different regularisation values C </B>

```python
# Higher C - More comples model to capture more details

logreg100 = LogisticRegression(C=1000, max_iter=100000).fit(X_train,y_train)
print("accuracy on train-set: {:.4f}".format(logreg100.score(X_train, y_train)))
print("accuracy on test-set : {:.4f}".format(logreg100.score(X_test, y_test)))
```

```python
# Lower C - Less complex model 

logreg001 = LogisticRegression(C=0.0001, max_iter=10000).fit(X_train,y_train)
print("accuracy on train-set: {:.4f}".format(logreg001.score(X_train, y_train)))
print("accuracy on test-set : {:.4f}".format(logreg001.score(X_test, y_test)))
```

Remember: The best value that we got a KNN was an Accurracy of 0.94! So linear regression seems is now a little better.

**Plot the coefficients**

```python
plt.figure(figsize=(10,6))

plt.plot(logreg.coef_.T, 'o', label="C=1")          # First experiment
plt.plot(logreg100.coef_.T, '^', label="C=1000")    # second experment
plt.plot(logreg001.coef_.T, 'v', label="C=0.0001")    # third experiment

plt.xticks(range(cancer.data.shape[1]), cancer.feature_names, rotation=90)
plt.hlines(0,0,cancer.data.shape[1])
plt.ylim(-5,5)
plt.xlabel('Index of coefficient')
plt.ylabel('Value of coefficient')
plt.legend()
plt.show()
print("done")

```
