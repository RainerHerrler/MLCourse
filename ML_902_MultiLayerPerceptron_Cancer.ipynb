# MLP with the Breast Cancer Data

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer

cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, random_state=0)
print("Done")
```

```python
import pandas as pd
data = pd.DataFrame(cancer.data, columns=[cancer.feature_names])
display(data)
```

```python
cancer.target
```

```python
from sklearn.neural_network import MLPClassifier
#MLPClassifier?
```

### Instantiate Algorithm and train it 

```python
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(
    random_state=42)
mlp.fit(X_train, y_train) 
print("Score on Training Data: {:.2f}".format(mlp.score(X_train, y_train)))
print("Score on Test Data:     {:.2f}".format(mlp.score(X_test, y_test)))
```

### Improve the results

MLP's are also sensible to scaling of the data. We can use the MinMaxScaler or the StandardScaler to bring features to the same range

```python
## First analyze and vizualize the data
import matplotlib.pyplot as plt 
X_train.shape
plt.plot(X_train.min(axis=0),'o', label = "min")
plt.plot(X_train.max(axis=0),'^', label = "max")
plt.legend(loc=1)
plt.xlabel("Feature Index")
plt.xlabel("Feature Range")
plt.xticks(range(30), cancer.feature_names, rotation=90)
plt.yscale("log")
print ("Done")
```

**Rescale Parameters with Standard Scaler**

```python
from sklearn.preprocessing import StandardScaler 
from sklearn.preprocessing import MinMaxScaler 

scaler = StandardScaler()
scaler.fit(X_train)                         #The data is presented to learn the distribution of each feature
X_train_scaled = scaler.transform(X_train)  #The learned scaling parameters are applied to the data
X_test_scaled = scaler.transform(X_test)

print("Mean:", X_train_scaled.mean(axis=0))
print("Std:", X_train_scaled.std(axis=0))
print("Max", X_train_scaled.max(axis=0))
print("Min", X_train_scaled.min(axis=0))
```

```python
import matplotlib.pyplot as plt 
X_train.shape
plt.plot(X_train_scaled.min(axis=0),'^', label = "min")
plt.plot(X_train_scaled.max(axis=0),'v', label = "max")
plt.plot(X_train_scaled.mean(axis=0),'o', label = "mean")

plt.legend(loc=1)
plt.xlabel("Feature Index")
plt.xlabel("Feature Range")
plt.xticks(range(30), cancer.feature_names, rotation=90)
print ("Done")
```

**Train & Evaluate rescaled data**

```python
mlp = MLPClassifier(random_state=42)

mlp.fit(X_train_scaled, y_train) 
print("Score on Training Data: {:.3f}".format(mlp.score(X_train_scaled, y_train)))
print("Score on Test Data:     {:.3f}".format(mlp.score(X_test_scaled, y_test)))
```

Rescaling the input helped. This was better about 4-5%. We may not forget to rescale also unseen data in the future

**We remove the warning by raising max_iter**

```python
mlp = MLPClassifier(
    random_state=42,
    max_iter=1000
)
mlp.fit(X_train_scaled, y_train) 
print("Score on Training Data: {:.3f}".format(mlp.score(X_train_scaled, y_train)))
print("Score on Test Data:     {:.3f}".format(mlp.score(X_test_scaled, y_test)))
print(mlp.alpha)
```

Observation: We raised accuracy on training data but not on test data. There is a significant gap between test and training. We need a more simple modell.

**Regularizing the model with alpha**

```python
MLPClassifier?
```

```python
mlp = MLPClassifier(
    random_state=42,
    max_iter=1000,
    alpha = 1 #default is 0.0001
)
mlp.fit(X_train_scaled, y_train) 
print("Score on Training Data: {:.3f}".format(mlp.score(X_train_scaled, y_train)))
print("Score on Test Data:     {:.3f}".format(mlp.score(X_test_scaled, y_test)))
```
