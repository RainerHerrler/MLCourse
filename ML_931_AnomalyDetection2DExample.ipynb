**Creating an Example**

```python
%matplotlib inline
from sklearn.datasets import make_moons
from sklearn.datasets import make_circles
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import numpy as np
import mglearn

X,y = make_blobs(n_samples=20, centers=1, n_features=2,cluster_std=0.5, random_state=0)
plt.figure(figsize=(8,6))
mglearn.discrete_scatter(X[:,0], X[:,1], np.ones(len(X)))
plt.ylim((2,7))
plt.xlim((-1,3))
plt.title("Original 2D data")
plt.show()

X,y = make_blobs(n_samples=20, centers=1, n_features=2,cluster_std=0.5, random_state=0)
plt.figure(figsize=(8,1))
mglearn.discrete_scatter(X[:,0], X[:,1]*0, np.ones(len(X)))
plt.title("Projection to X-Axis")
plt.ylim((-1,1))
plt.xlim((-1,3))
plt.show()

X,y = make_blobs(n_samples=20, centers=1, n_features=2,cluster_std=0.5, random_state=0)
plt.figure(figsize=(1,6))
mglearn.discrete_scatter(X[:,0]*0, X[:,1], np.ones(len(X)))
plt.title("Projection to Y-Axis")
plt.ylim((2,7))
plt.xlim((-1,1))
plt.show()






print("Done")
```

**Calculate Model**

```python
import numpy as np
mean0 = np.mean(X[:,0])
mean1 = np.mean(X[:,1])
std0 = np.std(X[:,0])
std1 = np.std(X[:,1])

import scipy.stats as stats
def p(x0,x1):
 return (
  stats.norm.pdf(x0, mean0, std0)*
  stats.norm.pdf(x1, mean1, std1))


```

**Print Model Parameters**

```python
print("Mean0", mean0)
print("Std0 ", std0)
print("Mean1", mean1)
print("Std1 ", std1)
```

**Plot Model (Probability Function) in 3D**

```python
%matplotlib notebook
fig = plt.figure(figsize = (12,10))
ax = plt.axes(projection='3d')

x = np.arange(-1, 3, 0.05)
y = np.arange(2, 7, 0.05)


XM, YM = np.meshgrid(x, y)
form = XM.shape
Z = p(XM.ravel(),YM.ravel())
Z = Z.reshape(form)

ax.plot_surface(XM, YM, Z, cmap = plt.cm.cividis)

# Set axes label
ax.set_xlabel('x0', labelpad=20)
ax.set_ylabel('x1', labelpad=20)
ax.set_zlabel('p(x)', labelpad=20)

plt.show()
```
