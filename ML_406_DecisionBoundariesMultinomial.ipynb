<h1>OneVersusRest and OneVersusAll Classifiers</h1>

**Loading and showing a new artificial dataset**

```python
from thwsplotting import * # self provided python for often used functions
```

```python
from sklearn import datasets
import matplotlib.pyplot as plt
import numpy as np
from mlxtend.plotting import plot_decision_regions

X, y = datasets.make_blobs(random_state=42)
plotScatter(X,y)
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
plt.legend(["Category 0","Category 1","Category 2"])
```

**Instanciate and train the model**

```python
from sklearn.svm import LinearSVC
from sklearn.multiclass import OneVsRestClassifier

svc = LinearSVC(multi_class='ovr',dual=False)
#svc = OneVsRestClassifier(LinearSVC()).fit(X,y)
svc.fit(X,y)
print("Score is:", svc.score(X,y))
```

```python
# Show the model
print("Coefficients: ", svc.coef_.shape)
print("Intercept", svc.intercept_.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier

svc = LogisticRegression(dual=False)
#svc = OneVsRestClassifier(LinearSVC()).fit(X,y)
svc.fit(X,y)
print("Score is:", svc.score(X,y))
# Show the model
print("Coefficients: ", svc.coef_.shape)
print("Intercept", svc.intercept_.shape)
```

**Plotting Decision Borders of individual hypothesis functions**

```python
plotScatter(X,y) # draws all poo
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")

line = np.linspace(np.min(X[:,0]),np.max(X[:,0]))
for coef, intercept, color in zip(svc.coef_, svc.intercept_, ['b','r', 'g']):
    plt.plot(line, -(line*coef[0]+intercept)/coef[1], c=color)
plt.ylim(np.min(X[:,1])-1,np.max(X[:,1])+1)
plt.legend(["Category 0","Category 1","Category 2", "Decision border 0", "Decision border 1", "Decision border 2"], loc=(1.01,0.3))
```

**Plotting Decision Borders of joint hypothesis function**

```python
fig, ax = plt.subplots(1, 1, figsize=(10, 7))
plot_decision_regions(X, y,clf=svc,ax=ax, scatter_kwargs={'s': 100});
plt.xlabel("Feature 0")
plt.ylabel("Feature 1")
line = np.linspace(-15,15)
for coef, intercept, color in zip(svc.coef_, svc.intercept_, ['b','r', 'g']):
   plt.plot(line, -(line*coef[0]+intercept)/coef[1], c=color)
plt.legend(["Category 0","Category 1","Category 2", "Decision border 0", "Decision border 1", "Decision border 2"], loc=(1.01,0.3))
```
