# Solving FourBlobs with a Gaussian Naive Bayes Model

**Create Data and show**

```python
from thwsplotting import *
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

X,y = make_blobs(centers=4, random_state=8)
#X,y = make_blobs(centers=8, random_state=7)
y=y%2
plotScatter(X,y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

**Train a Gaussian Naive Bayes Classifier **

```python
from mlxtend.plotting import plot_decision_regions
from sklearn.naive_bayes import GaussianNB


clf = GaussianNB().fit(X,y)
#clf = GaussianNB(var_smoothing=1e-1).fit(X,y) # tweaking parameters 
fig = plt.figure(figsize=(6,4))
plot_decision_regions(X, y,  markers='o^', clf=clf, hide_spines=False)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
```

**Exercise: Vizualize data and train a Naive Bayes Classifier**

```python
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
import numpy as np

X,y = make_blobs(centers=1, n_samples=300, random_state=8)
y = np.logical_xor(X[:, 0] > 7, X[:, 1] > 9.5)*1

plotScatter(X,y)
print("Done")
```

**Solution**

```python
from mlxtend.plotting import plot_decision_regions
from sklearn.naive_bayes import GaussianNB

plotScatter(X, y)
plt.title("Blob with XOR")
plt.show()


```

```python
clf = GaussianNB().fit(X,y)
#clf = GaussianNB(var_smoothing=1e-1).fit(X,y)
fig = plt.figure(figsize=(6,4))
plot_decision_regions(X, y,  markers='o^', clf=clf, hide_spines=False)
plt.show()

clf.score(X,y) #Score auf den Trainingsdaten
```

```python
from sklearn.neighbors import KNeighborsClassifier

clf = KNeighborsClassifier(n_neighbors = 7).fit(X,y)

fig = plt.figure(figsize=(6,4))
plot_decision_regions(X, y,  markers='o^', clf=clf, hide_spines=False)
plt.show()

clf.score(X,y)
```
