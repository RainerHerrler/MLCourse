# Artificial Neural Networks in SKLearn with the 2 Moons dataset

```python
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_moons
from thwsplotting import *

import matplotlib.pyplot as plt
import numpy as np

X,y = make_moons(n_samples = 100, noise =0.2, random_state=3)
X_train, X_test, y_train, y_test = train_test_split(X,y, stratify=y, random_state=42)
plotScatter(X_train, y_train)
print("Done")
```

**A first try to learn with an ANN** 

```python
from sklearn.neural_network import MLPClassifier
#MLPClassifier?
```

```python
from sklearn.neural_network import MLPClassifier
mlp = MLPClassifier(
    activation='logistic',       # sigmoid is called "logistic" here
    hidden_layer_sizes=[5,5],    # one lyer with 100 neurons, this was called s in the slides
    learning_rate_init=0.4,      # the learning rate for gradient descent 
    max_iter=10000,              # 100 -> 1000 max number of steps for gradient descent
    random_state=0)
mlp.fit(X_train, y_train) # training 
```

```python
print("Score on Training", mlp.score(X_train, y_train))
print("Score on Test", mlp.score(X_test, y_test))
```

```python
# Test new Points 
newPoint=np.array([[1,0.25]])
print("Probabilities for blue/orange: ", mlp.predict_proba(newPoint))
print("Predicted Label:", mlp.predict(newPoint))
```

```python
from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y, clf=mlp, hide_spines=False)
```

```python
import matplotlib.pyplot as plt
import networkx as nx

# ---- Build graph ----

clf = mlp
layers = [X.shape[1]] + list(clf.hidden_layer_sizes) + [clf.n_outputs_]

G = nx.DiGraph()

pos = {}
node_id = 0
layer_nodes = []

# Create nodes
for i, layer_size in enumerate(layers):
    current_layer = []
    
    for j in range(layer_size):
        G.add_node(node_id)
        pos[node_id] = (i, -j)
        current_layer.append(node_id)
        node_id += 1
        
    layer_nodes.append(current_layer)

# Create edges
for i in range(len(layer_nodes)-1):
    for n1 in layer_nodes[i]:
        for n2 in layer_nodes[i+1]:
            G.add_edge(n1, n2)

# ---- Plot ----

plt.figure(figsize=(10, 6))

nx.draw(
    G,
    pos,
    with_labels=False,
    node_size=800,
    arrows=False
)

plt.title("MLPClassifier Neural Network Structure")
plt.axis("off")
plt.show()
```
