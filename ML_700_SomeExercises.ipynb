# Various problems with features

**Problem 1: Binary Classification with 4 Blobs**

```python
%matplotlib inline
import mglearn
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

X,y = make_blobs(centers=3, random_state=2)
y=y%2 # from 3 classes we reduce to tweo
mglearn.discrete_scatter(X[:,0],X[:,1],y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

**Problem 2: Multinomial Classification with 4 Blobs**

```python
X,y = make_blobs(centers=4, random_state=8)
print(X.shape)
mglearn.discrete_scatter(X[:,0],X[:,1],y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

**Problem 3 : Checkers**

```python
from sklearn.datasets import make_blobs
import numpy as np

np.random.seed(0)
X = np.random.rand(200,2)*10
y = np.logical_xor(X[:, 0] > 5, X[:, 1] > 5)*1
mglearn.discrete_scatter(X[:,0],X[:,1],y)
```

**Problem 4 : Trend**

```python
np.random.seed(3)
X = np.random.rand(20,1)*10
y = np.log(X[:, 0]*2)+ (np.random.rand(20)*1.5)
#y = np.exp(y)

mglearn.discrete_scatter(X[:,0],y)
plt.xlabel('Feature')
plt.ylabel('Label')
```
