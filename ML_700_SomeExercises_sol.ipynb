# Various problems with features

**Problem 1: Binary Classification with 4 Blobs**

```python
%matplotlib inline
import mglearn
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

X,y = make_blobs(centers=3, random_state=2)
y=y%2
mglearn.discrete_scatter(X[:,0],X[:,1],y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

svc = LinearSVC(dual="auto")
svc.fit(X, y)

# explcitit score calculation
predictions = svc.predict(X);
score = accuracy_score(y, predictions);

#shortcut
print("Score on Training", svc.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^', clf=svc, hide_spines=False)
print(svc.coef_)
```

**Problem 2: Multinomial Classification with 4 Blobs**

```python
X,y = make_blobs(centers=4, random_state=8)
mglearn.discrete_scatter(X[:,0],X[:,1],y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier().fit(X,y)
print("Training Score ", knn.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^sv', clf=knn, hide_spines=False)
```

```python
from sklearn.naive_bayes import GaussianNB

gauss = GaussianNB().fit(X,y)
print("Training Score ", gauss.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^sv', clf=gauss, hide_spines=False)
```

```python
#LogisticRegression?
```

```python
from sklearn.linear_model import LogisticRegression
logreg = LogisticRegression(multi_class="multinomial").fit(X,y)
print("Training Score ", logreg.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^sv', clf=logreg, hide_spines=False)
```

**Problem 3 : Checkers**

```python
from sklearn.datasets import make_blobs
import numpy as np
import mglearn

np.random.seed(0)
X = np.random.rand(200,2)*10
y = np.logical_xor(X[:, 0] > 5, X[:, 1] > 5)*1
mglearn.discrete_scatter(X[:,0],X[:,1],y)


```

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier(max_depth=4).fit(X,y)
print("Training Score ", clf.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^sv', clf=clf, hide_spines=False)
```

```python
from sklearn.tree import DecisionTreeClassifier

clf = DecisionTreeClassifier().fit(X,y)
print("Training Score ", clf.score(X,y))

from mlxtend.plotting import plot_decision_regions
plot_decision_regions(X, y,  markers='o^sv', clf=clf, hide_spines=False)
```

**Problem 4 : Trend**

```python
import numpy as np

np.random.seed(3)
X = np.random.rand(20,1)*10
y = np.log(X[:, 0])*2 + (np.random.rand(20)*1.5)

mglearn.discrete_scatter(X[:,0],y)
```

```python
from sklearn.tree import DecisionTreeRegressor

clf = DecisionTreeRegressor(max_depth=4).fit(X,y)
print("Training Score ", clf.score(X,y))
```

```python
# Create Test data in the same way
X_test = np.random.rand(20,1)*10
y_test = np.log(X_test[:, 0])*2+ (np.random.rand(20)*1.5)
print("Test Score ", clf.score(X_test,y_test))

```

```python
#Compare ground truth vs prediction 
mglearn.discrete_scatter(X[:,0],y)   # original data
mglearn.discrete_scatter(X_test[:,0],y_test, c='yellow') # ground truth values

y_predict = clf.predict(X_test)
mglearn.discrete_scatter(X_test[:,0],y_predict, c='red') # made predictions
```
