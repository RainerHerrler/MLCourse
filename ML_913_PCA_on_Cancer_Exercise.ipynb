# PCA on the Cancer-Dataset

This exercise starts with the breast-cancer dataset and asks you to reduce the feature space with PCA before comparing results.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler
import numpy as np
import matplotlib.pyplot as plt

cancer = load_breast_cancer()

scaler = StandardScaler()
scaler.fit(cancer.data)
X = scaler.transform(cancer.data)
y = cancer.target
print("Data dimensionality", X.shape)
```
