# Predicting RAM Prices



```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

ram_prices = pd.read_csv("ram_price.csv")
display(ram_prices)
```

**Plotting the data**

We use the function "semilogy" from matplotlib to create a plot with logarithmic scale, you can also check how the plot would look like if you habe regular plot.

```python
plt.figure(figsize=(10,7))
plt.semilogy(ram_prices.date,ram_prices.price, marker='o')
#plt.plot(ram_prices.date,ram_prices.price, marker='o')
plt.xlabel("Year")
plt.ylabel("Price in $/Mbyte")
```

**Split in Training and Test Data**

```python
data_train = ram_prices[ram_prices.date <2000]  # train with hisoric data before
data_test = ram_prices[ram_prices.date >=2000]  # predict the future

# print("Before conversion", data_train.date.to_numpy().shape)
# Think about the required shape for the typical fit function
```

```python
X_train = data_train.date.to_numpy().reshape(-1,1)
y_train = data_train.price.to_numpy()
X_test = data_test.date.to_numpy().reshape(-1,1)
y_test = data_test.price.to_numpy()
X_all = ram_prices.date.to_numpy().reshape(-1,1)
Y_all = ram_prices.price.to_numpy()
print("Done")

print("Samples before 2000",y_train.shape)
print("Samples after 2000",y_test.shape)
print("Sum of samples",Y_all.shape)
```

**Comparing Linear Regression and Decision-Trees**

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression;

tree = DecisionTreeRegressor().fit(X_train, y_train)
linear_reg = LinearRegression().fit(X_train, y_train)

#Predict everything
pred_tree = tree.predict(X_train)
pred_lr = linear_reg.predict(X_train)

print('Decision Tree     Training Score:{:.2f}'.format(tree.score(X_train, y_train)))
print('Linear Regression Training Score:{:.2f}'.format(linear_reg.score(X_train, y_train)))
```

<details>
<summary>Click for Explanation</summary>
The linear model cannot find a model because the line we see is on a log scale, so actually the  relation between time and price is not a linear relationship.

The decision tree however if complex enough and unrestrited in depth can fully explain the training data. 
</details>

```python
print('Decision Tree     Test Score:{:.2f}'.format(tree.score(X_test, y_test)))
print('Linear Regression Test Score:{:.2f}'.format(linear_reg.score(X_test, y_test)))
```

<details>
<summary>Click for Explanation</summary>
Even though the decision tree has perfect score on training data, it terribly fails on test data because it just predicts the closest point (the last price). It is unable to extrapolate to future prices. It could predict past prices because its able to intrapolate (average) prices between existing datapoints.

The linear regression couldnt predict in training and cannot predict on test data.
</details>

**Vizualize Prediction on Training data**

```python
plt.figure(figsize=(10,7))
plt.semilogy(data_train.date, data_train.price, linestyle=(6, (5.0, 7.0)), c='r', label='Training data')
plt.semilogy(data_train.date, pred_tree, linestyle=(0, (5.0, 7.0)), c='g', label='Decision Tree')
plt.semilogy(data_train.date, pred_lr, label='Linear Regression')
plt.legend()
```

**Vizualize Prediction on Complete Data (Training & Test)**

```python
plt.figure(figsize=(10,7))
plt.semilogy(X_all, Y_all, linestyle=(6, (5.0, 7.0)), c='r', label='Trainingsdaten')
plt.semilogy(X_all, tree.predict(X_all), linestyle=(0, (5.0, 7.0)), c='g', label='Decision Tree')
plt.semilogy(X_all, linear_reg.predict(X_all), label='Linear Regression')
plt.legend()
```

**Apply Logarithmic Scaling of the target value**

The linear regression model can be fixed by rescaling the data before we train the model.

```python
#log transform prediction labels
y_train = np.log(y_train)

tree = DecisionTreeRegressor().fit(X_train, y_train)
linear_reg = LinearRegression().fit(X_train, y_train)

#Predict everything
pred_tree = tree.predict(X_all)
pred_lr = linear_reg.predict(X_all)

print('Decision Tree     Test Score:{:.2f}'.format(tree.score(X_test, np.log(y_test))))
print('Linear Regression Test Score:{:.2f}'.format(linear_reg.score(X_test, np.log(y_test))))

#undo log transformation
pred_tree = np.exp(pred_tree)
pred_lr = np.exp(pred_lr)
y_train = np.exp(y_train)



```
