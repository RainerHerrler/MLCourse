# Step by step calculation of the K-Means

**Creating data derived from iris**<br>
Normalizing is recommended if you have values of different units

```python
%matplotlib inline
import pandas as pd
import numpy as np
import mglearn
import matplotlib as mpl
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# loading original IRIS data (4 Feature)
iris = datasets.load_iris()

# Reducing the feature data to the last 2 columns 
X = iris.data[:,2:4] 
X = MinMaxScaler().fit_transform(X) # column-wise rescaling to squish the contents between 0 and 1
#X = StandardScaler().fit_transform(X)
print("Shape", X.shape)
```

**Helper functions**

```python
def euclidean(point, data):
    """
    Euclidean distance between point & data.
    Point has dimensions (m,), data has dimensions (n,m), and output will be of size (n,).
    """
    return np.sqrt(np.sum((point - data)**2))
print("Distance function defined")
```

```python
def plotCluster(X, y, centroids):
    """
    Plots X conataining two clusters (labeled 0,1), and centroid points (labeled 2,3)
    """
    #colors blue and red
    cm_cycle = ['#7070ff', '#ff5050', '#7070ff', '#ff5050' ]

    Xplot = np.vstack((X,centroids))
    yplot = np.hstack((y,np.arange(len(centroids))%2+2))

    plt.figure(figsize=(12,9))
    mglearn.discrete_scatter(Xplot[:, 0], Xplot[:, 1], yplot, markers=['o','o','X','X'], c=cm_cycle)
    plt.legend(['Cluster 1', 'Cluster 2', 'Centroid 1', 'Centroid 2'])
    plt.show()
print("Plotting function defined")
```

**Step 1: Plot data and create two random centroids**

```python
centroids=np.array([[0.3,0.75],
                   [0.75,0.3]])
y= np.array(np.ones(len(X))) # all the same class
y[0]=0

plotCluster(X,y,centroids)
```

**Step 2: Assign all samples to closest cluster center** 

```python
y= np.array([euclidean(point, centroids[0])>euclidean(point,centroids[1]) for point in X])
plotCluster(X,y,centroids)
```
