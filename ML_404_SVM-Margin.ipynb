<h1>SVM Representation - Line with margin</h1>

```python
%matplotlib inline 
import numpy as np
import math
import matplotlib.pyplot as plt
import matplotlib as mpl
import math

xmin=-1
xmax=20

def plotBoundaryWithMargin(theta): # theta consists of three coefficients 
  x = np.linspace(xmin,xmax)
  y    = (theta[1]*x+theta[0])/theta[2]     # the decision boundary
  ytop = (theta[1]*x+theta[0]+1)/theta[2]   # the upper margin 
  ybtm = (theta[1]*x+theta[0]-1)/theta[2]   # the lower margin
  plt.figure(figsize=(8,6))
  label='{}𝑥+{}𝑦+(−{})='.format(theta[1], theta[0], theta[2])
  plt.plot(x,ytop, linewidth=2, color='blue', linestyle = "dashed",  label='${}+1$'.format(label)) # upper curb / obenrer Randstein
  plt.plot(x,y, linewidth=2, color='green', label='${}0$'.format(label))  # center line / decision boundary
  plt.plot(x,ybtm, linewidth=2, color='blue',  linestyle = "dashed", label='${}-1$'.format(label)) #lower curb / unteren Randstein
  plt.hlines(y=0, xmin=xmin, xmax=xmax, linewidth=1,color='black')
  plt.vlines(x=0, ymin=-5, ymax=20, linewidth=1, color='black')
  plt.ylim(-5,10)
  plt.xlim(xmin,20)
  plt.box(False)
  plt.legend(prop={'size': 16})
  plt.show()

theta = np.array([1.6, -0.2,0.3])
plotBoundaryWithMargin(theta)

```
