# MLPRegressor on the California housing dataset

```python
# a lot of imports
%matplotlib inline
import sklearn
import mglearn
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# We used The California Housing dataset already before, the first time with KNN

from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
```

```python
print("Samples x features:", housing.data.shape)
```

```python
X = StandardScaler().fit_transform(housing.data)
y = housing.target 

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
print("Training data shape:", X_train.shape)
print("Training data shape:", X_test.shape)
```

## Lets find something out about the dataset

<b>Lasso Regression - L1-"Regularization"</b>

```python
from sklearn.linear_model import Lasso
# we increase the default setting of "max_iter",
# otherwise the model would warn us that we should increase max_iter.
lasso001 = Lasso(alpha=0.001, max_iter=10000).fit(X_train, y_train)
print("Training set score: {:.2f}".format(lasso001.score(X_train, y_train)))
print("Test set score: {:.2f}".format(lasso001.score(X_test, y_test)))
print("Number of features used:", np.sum(lasso001.coef_ != 0))
```

**Using a neural net**

```python
from sklearn.neural_network import MLPRegressor
MLPRegressor?
```

```python
from sklearn.neural_network import MLPRegressor
net = MLPRegressor(
    activation = 'relu', 
    hidden_layer_sizes = (100,100,), 
    max_iter = 400, 
    alpha = 0.01, # not the learning rate but for regularisation
    random_state=42,
    verbose = True
)
#splits data and creates a basic neural network
net.fit(X_train, y_train)
print('R2 score Training is: {}'.format(net.score(X_train, y_train)))
print('R2 score Test is:     {}'.format(net.score(X_test, y_test)))
```

```python
y_prediction = net.predict(X_test)
dict = {'Actual Values' : y_test*100000, 'Predicted Values' : y_prediction*100000}
final_df = pd.DataFrame(dict)
styled_df = final_df.head(10).style.format({
    'Actual Values': "${:,.0f}",
    'Predicted Values': "${:,.0f}"
})
display(styled_df)
```
