# OneVersusRest and OneVersusOne Classifiers

**Load the irdis dataset**

```python
from sklearn import datasets
X, y = datasets.load_iris(return_X_y=True) # shortcut without the additional information (labelnames etc)
print("Features:",X.shape)
print("Ground Truth Labels:",y)
```

**Wrapp the OneVsRest around the LinearSVC and train**

```python
from sklearn import datasets
from sklearn.multiclass import OneVsOneClassifier
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression

#Wrapp the OneVsRest around the LinearSVC (Other combinations possible OneVsOneClassifier(LogisticRegression))
clf = OneVsRestClassifier(LogisticRegression(C=1, random_state=0,dual=False))
clf.fit(X, y)
```

```python
pred = clf.predict(X)
print("Predictions:\n", pred)
#print("Training-Score:", clf.score(X,y))
```

```python
#Wrapp the OneVsRest around the LinearSVC (Other combinations possible OneVsOneClassifier(LogisticRegression))
clf2 = OneVsOneClassifier(LogisticRegression(C=1, random_state=0,dual=False))
clf2.fit(X, y)
pred = clf2.predict(X)
print("Predictions:\n", pred)
print("Training-Score:", clf2.score(X,y))
```

```python
# lets try directly
X, y = datasets.load_iris(return_X_y=True)
clf = LogisticRegression(C=1, random_state=0, max_iter=1000)
clf.fit(X, y)
pred = clf.predict(X)
print("Predictions:\n", pred)
print("Score:", clf.score(X,y))
```

```python
X, y = datasets.load_iris(return_X_y=True)
clf = LinearSVC(C=1, multi_class='ovr', random_state=0,dual=False)
clf.fit(X, y)
pred = clf.predict(X)
print("Predictions:\n", pred)
print("Score:", clf.score(X,y))
```
