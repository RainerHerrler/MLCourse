```python
#On windows you have to install graphvoz before https://graphviz.org/download/
# and pip install graphviz
import matplotlib.pyplot as plt
import graphviz
import imageio.v2 as imageio
from imageio.v2 import imread

def plot_animal_tree(ax=None):
    if ax is None:
        ax = plt.gca()
    mygraph = graphviz.Digraph(node_attr={'shape': 'box','fontname' :'Candara'},
                               edge_attr={'labeldistance': "10.5", 'fontname' :'Candara'},
                               format="png",
                              )
    mygraph.graph_attr['size'] = '500,500'
    mygraph.graph_attr['resolution'] = '200'
    mygraph.node("0", "Has feathers?")
    mygraph.node("1", "Can fly?")
    mygraph.node("2", "Has fins?")
    mygraph.node("3", "Hawk")
    mygraph.node("4", "Penguin")
    mygraph.node("5", "Dolphin")
    mygraph.node("6", "Bear")
    mygraph.edge("0", "1", label="True")
    mygraph.edge("0", "2", label="False")
    mygraph.edge("1", "3", label="True")
    mygraph.edge("1", "4", label="False")
    mygraph.edge("2", "5", label="True")
    mygraph.edge("2", "6", label="False")
    mygraph.render("tmp")
    ax.imshow(imread("tmp.png"), resample=False)
    ax.set_axis_off()
    
plt.figure(figsize=(16, 16))
plot_animal_tree()
```
