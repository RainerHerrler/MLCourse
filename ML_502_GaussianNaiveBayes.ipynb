# Solving FourBlobs with a Gaussian Naive Bayes Model

**Create Data and show**

```python
from thwsplotting import *
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

X,y = make_blobs(centers=4, random_state=8)
#X,y = make_blobs(centers=8, random_state=7)
y=y%2
plotScatter(X,y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

**Train a Gaussian Naive Bayes Classifier**

```python
#Install this if you execute it for the first time
#!pip install mlxtend
```

```python
from mlxtend.plotting import plot_decision_regions
from sklearn.naive_bayes import GaussianNB

clf = GaussianNB().fit(X,y)
#clf = GaussianNB(var_smoothing=1e-1).fit(X,y) # tweaking parameters 
fig = plt.figure(figsize=(6,4))
plot_decision_regions(X, y,  markers='o^', clf=clf, hide_spines=False)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
```

**Multiple Classes**

```python
X,y = make_blobs(centers=4, random_state=8)
plotScatter(X,y)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
clf = GaussianNB().fit(X,y)
#clf = GaussianNB(var_smoothing=1e-1).fit(X,y) # tweaking parameters 
fig = plt.figure(figsize=(6,4))
plot_decision_regions(X, y,  markers='o^vs', clf=clf, hide_spines=False)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
```
